# 02 · Music / sound-effect removal

Vocal stem with **audio-separator** (`separation.model`). The segmenter later measures the music level of each clip and keeps the **original audio** wherever there is (almost) no music, so separation artifacts only exist where separation was needed.

> Off by default (`separation.mode: never`): clean podcasts are used as they are. Set `auto` for sources with music beds.

```python
import os, sys, json
from pathlib import Path
REPO = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(REPO)); os.chdir(REPO)
if Path("/workspace").exists():
    os.environ.setdefault("HF_HOME", "/workspace/hf_cache")
import numpy as np, pandas as pd
import plotly.express as px
from egyspeech import nb
cfg, lay = nb.cfg_and_layout()          # configs/config.yaml
print("work dir:", lay.root)
```

```python
nb.run("separate")
```

```python
from egyspeech.io import decode_audio, read_audio, resample
from IPython.display import Audio, display
vid = next(v for v in pd.DataFrame(nb.read_jsonl(lay.videos))["id"] if lay.vocals(v).exists())
raw = decode_audio({v["id"]: v for v in nb.read_jsonl(lay.videos)}[vid]["path"], 24000)
voc, sr = read_audio(lay.vocals(vid))
start = int(min(120, len(raw) / 24000 / 2) * 24000)
print("original"); display(Audio(raw[start:start + 20 * 24000], rate=24000))
print("vocal stem"); display(Audio(voc[start:start + 20 * 24000], rate=24000))
```
